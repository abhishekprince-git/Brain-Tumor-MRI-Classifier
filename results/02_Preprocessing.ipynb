```python
# Define Constants
IMG_SIZE = (224, 224) # Standard size for most CNNs (e.g., VGG, ResNet)
BATCH_SIZE = 32
SEED = 42
VALID_EXTENSIONS = ('.jpg', '.jpeg', '.png', '.bmp')

# Class Mapping (Binary Classification: Tumor vs Non-Tumor)
# Based on snippet: glioma, meningioma, pituitary -> Tumorous (1), notumor -> Non-Tumorous (0)
CLASS_MAPPING = {
    "notumor": 0,
    "glioma": 1,
    "meningioma": 1,
    "pituitary": 1
}
CLASS_NAMES = ['Non-Tumourous', 'Tumourous']
LABEL_MAP_REVERSE = {0: 'Non-Tumourous', 1: 'Tumourous'}

print("Setup Complete.")
```

```python
def load_dataset_to_df(base_dir, split_name):
    """
    Loads image paths and labels from directory structure into a DataFrame.
    """
    data = []
    print(f"Loading {split_name} dataset...")

    for class_name in os.listdir(base_dir):
        class_path = os.path.join(base_dir, class_name)

        if not os.path.isdir(class_path):
            continue

        label = CLASS_MAPPING.get(class_name, -1)
        if label == -1:
            print(f"Warning: Unknown class folder '{class_name}' skipped.")
            continue

        for img_file in os.listdir(class_path):
            if img_file.lower().endswith(VALID_EXTENSIONS):
                img_path = os.path.join(class_path, img_file)
                data.append({
                    'image_path': img_path,
                    'label': label,
                    'class_name': class_name,
                    'split': split_name
                })

    df = pd.DataFrame(data)
    print(f"Loaded {len(df)} images for {split_name}.")
    return df
```

```python
# Load Training Data
train_df = load_dataset_to_df(TRAIN_DIR, "train")

# Load Test Data
test_df = load_dataset_to_df(TEST_DIR, "test")

# Combine to check for duplicates across entire dataset before splitting validation
full_df = pd.concat([train_df, test_df], ignore_index=True)

print(f"Total images loaded: {len(full_df)}")
full_df.head()
```

```python
def compute_perceptual_hash(img_path):
    """Compute perceptual hash for an image."""
    try:
        with Image.open(img_path) as img:
            # Convert to grayscale for consistent hashing
            img_gray = img.convert('L')
            phash = imagehash.phash(img_gray)
            return str(phash)
    except Exception as e:
        return None

def find_duplicates_and_leakage(df):
    """
    Identifies exact duplicates (same hash) and near-duplicates (similar phash)
    to prevent data leakage between train/val/test.
    """
    print("Computing perceptual hashes... This may take a minute.")
    df['phash'] = df['image_path'].apply(compute_perceptual_hash)

    # Drop rows where hashing failed
    df = df.dropna(subset=['phash'])

    # Group by hash to find duplicates
    hash_groups = df.groupby('phash')['image_path'].apply(list).to_dict()

    duplicate_count = sum(1 for paths in hash_groups.values() if len(paths) > 1)
    print(f"Found {duplicate_count} groups of duplicate/near-duplicate images.")

    # Identify Leakage: Same hash appearing in both Train and Test (or Val)
    # We will handle this by keeping only one instance per hash group.

    # Create a mapping of hash -> first occurrence index to keep
    kept_indices = []
    removed_indices = []

    # Sort dataframe to ensure deterministic selection (e.g., keep first path alphabetically)
    df_sorted = df.sort_values(by='image_path')

    seen_hashes = set()

    for idx, row in df_sorted.iterrows():
        h = row['phash']
        if h not in seen_hashes:
            seen_hashes.add(h)
            kept_indices.append(idx)
        else:
            removed_indices.append(idx)

    print(f"Keeping {len(kept_indices)} unique images.")
    print(f"Removing {len(removed_indices)} duplicates/leaked images.")

    # Filter dataframe to keep only unique images
    df_clean = df.loc[kept_indices].copy()

    return df_clean, removed_indices
```

```python
# Apply duplicate removal
clean_df, removed_indices = find_duplicates_and_leakage(full_df)

# Update our train and test dfs based on the clean list
# Note: In a real scenario, you might want to re-split entirely.
# Here we filter the existing splits to remove the leaked/duplicate entries.
train_df = clean_df[clean_df['split'] == 'train'].reset_index(drop=True)
test_df = clean_df[clean_df['split'] == 'test'].reset_index(drop=True)

print(f"Clean Train Size: {len(train_df)}")
print(f"Clean Test Size: {len(test_df)}")
```

```python
# Split the cleaned training data into Train and Validation sets
# Stratify by label to maintain class distribution
train_final_df, val_df = train_test_split(
    train_df,
    test_size=0.2,
    random_state=SEED,
    stratify=train_df['label']
)

# Reset indices
train_final_df = train_final_df.reset_index(drop=True)
val_df = val_df.reset_index(drop=True)
test_df = test_df.reset_index(drop=True)

print(f"Final Training Set: {len(train_final_df)}")
print(f"Validation Set: {len(val_df)}")
print(f"Test Set: {len(test_df)}")

# Save splits to CSV for reproducibility
train_final_df.to_csv(os.path.join(RESULTS_DIR, "train_split.csv"), index=False)
val_df.to_csv(os.path.join(RESULTS_DIR, "val_split.csv"), index=False)
test_df.to_csv(os.path.join(RESULTS_DIR, "test_split.csv"), index=False)
```

```python
#  Class Distribution & Visualization
def plot_class_distribution(df, title):
    counts = df['label'].value_counts().sort_index()
    labels = [LABEL_MAP_REVERSE[i] for i in counts.index]

    plt.figure(figsize=(8, 5))
    bars = plt.bar(labels, counts.values, color=['#ff9999','#66b3ff'])

    # Add count labels on bars
    for bar in bars:
        yval = bar.get_height()
        plt.text(bar.get_x() + bar.get_width()/2, yval + 0.5, int(yval), ha='center', va='bottom')

    plt.title(title)
    plt.xlabel('Class')
    plt.ylabel('Count')
    plt.ylim(0, max(counts.values) * 1.1)
    plt.show()

plot_class_distribution(train_final_df, "Training Set Class Distribution")
plot_class_distribution(val_df, "Validation Set Class Distribution")
plot_class_distribution(test_df, "Test Set Class Distribution")
```

**Class Weights**

The binary dataset contains more Tumourous images than Non-Tumourous images.

Class weights are calculated automatically using the training-set distribution. The minority class receives a higher weight, while the majority class receives a lower weight.

During CNN training, these weights help reduce the effect of class imbalance and encourage the model to learn both classes rather than favoring the majority class.

```python
# Calculate class weights to handle imbalance
# The model will pay more attention to the minority class (Non-Tumourous)

y_train = train_final_df['label'].values

classes = np.unique(y_train)
class_weights_array = compute_class_weight(
    class_weight='balanced',
    classes=classes,
    y=y_train
)

class_weights = dict(zip(classes, class_weights_array))

print("Calculated Class Weights:")
for cls, weight in class_weights.items():
    print(f"{LABEL_MAP_REVERSE[cls]} (Class {cls}): {weight:.4f}")

# Save class weights
import json
with open(os.path.join(RESULTS_DIR, "class_weights.json"), 'w') as f:
    json.dump({str(k): v for k, v in class_weights.items()}, f)
```

```python
# Define Image Data Generators for Augmentation
# Only augment training data to prevent overfitting.
# Validation and Test data should only be rescaled.

# Training Generator with Augmentation
train_datagen = ImageDataGenerator(
    rescale=1./255,          # Normalize pixel values to [0, 1]
    rotation_range=20,       # Random rotation up to 20 degrees
    width_shift_range=0.2,   # Random horizontal shift
    height_shift_range=0.2,  # Random vertical shift
    shear_range=0.2,         # Shear intensity
    zoom_range=0.2,          # Random zoom
    horizontal_flip=True,    # Random horizontal flip
    fill_mode='nearest'      # Fill mode for new pixels
)

# Validation and Test Generator (Only Rescaling)
val_test_datagen = ImageDataGenerator(
    rescale=1./255
)

print("ImageDataGenerators configured.")
```

```python
# Helper function to create generators from DataFrames
def create_generator(df, datagen, batch_size, img_size, shuffle=True):
    return datagen.flow_from_dataframe(
        dataframe=df,
        x_col='image_path',
        y_col='label',
        target_size=img_size,
        batch_size=batch_size,
        class_mode='raw', # Returns labels as they are (0 or 1)
        shuffle=shuffle,
        seed=SEED
    )
```

```python
# Create Generators
train_generator = create_generator(train_final_df, train_datagen, BATCH_SIZE, IMG_SIZE, shuffle=True)
val_generator = create_generator(val_df, val_test_datagen, BATCH_SIZE, IMG_SIZE, shuffle=False)
test_generator = create_generator(test_df, val_test_datagen, BATCH_SIZE, IMG_SIZE, shuffle=False)

print("Generators created successfully.")
print(f"Steps per epoch (Train): {train_generator.samples // train_generator.batch_size}")
print(f"Steps per epoch (Val): {val_generator.samples // val_generator.batch_size}")
```

```python
# Visualize some augmented images to verify preprocessing
def plot_augmented_samples(generator, num_samples=9):
    plt.figure(figsize=(15, 15))

    # Get a batch of images
    images, labels = next(generator)

    for i in range(num_samples):
        plt.subplot(3, 3, i + 1)
        plt.imshow(images[i])
        title = LABEL_MAP_REVERSE[int(labels[i])]
        plt.title(title, fontsize=10)
        plt.axis('off')

    plt.suptitle("Augmented Training Samples", fontsize=16)
    plt.tight_layout()
    plt.show()

# Reset generator to start from beginning
train_generator.reset()
plot_augmented_samples(train_generator)
```
